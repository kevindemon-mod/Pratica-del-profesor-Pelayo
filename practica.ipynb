```python
class Matricula:
    def costo(self):
        return 60.0

class Extemporanea(Matricula):
    def costo(self):
        return super().costo() + 15.0

class Desaprobado(Matricula):
    def costo(self) -> float:
        return super().costo() + 15.0

class Becado(Matricula):
    def costo(self):
        return super().costo() / 2

# MRO: Becado -> Desaprobado -> Extemporánea -> General
# Operación matemática: (((60 + 15) + 15) / 2) = 45.0
class Combinada1(Becado, Desaprobado, Extemporanea):
    pass

# MRO: Desaprobado -> Extemporánea -> Becado -> General
# Operación matemática: ((60 / 2) + 15) + 15 = 60.0
class Combinada2(Desaprobado, Extemporanea, Becado):
    pass

if __name__ == "__main__":
    luis = Combinada1()
    rosa = Combinada2()

    print(f"Pago de Luis (Programador 1): S/ {luis.costo()}")
    print(f"Pago de Rosa (Programador 2): S/ {rosa.costo()}")

    try:
        class FallaEstructural(Matricula, Becado):
            pass
    except TypeError as e:
        print(f"\nError por orden lógico")
        print(f"Python bloquea la ejecución lanzando TypeError: {e}")
```

```python
class Matricula:
    def __init__(self, estudiante):
        self._estudiante = estudiante
        self._costo_base = 60.0
        self._recargos = []
        self._tiene_beca = False

    @property
    def estudiante(self):
        return self._estudiante

    @property
    def costo_base(self):
        return self._costo_base

    @property
    def tiene_beca(self):
        return self._tiene_beca

    def aplicar_beca(self):
        self._tiene_beca = True

    def agregar_recargo(self, monto, motivo):
        if monto > 0:
            self._recargos.append((monto, motivo))
        else:
            raise ValueError("El monto del recargo debe ser mayor a 0.")

    @property
    def costo_final(self):
        """
        Calcula dinámicamente el total: suma todos los recargos al
        costo base y aplica el descuento de la beca al final.
        """
        total = self._costo_base

        for monto, _ in self._recargos:
            total += monto

        if self._tiene_beca:
            total = total / 2
        return total

if __name__ == "__main__":

    luis = Matricula("Luis Huamaní Cárdenas")
    luis.agregar_recargo(15.0, "Extemporánea")
    luis.agregar_recargo(15.0, "Índice Desaprobatorio")
    luis.aplicar_beca()

    print(f"Estudiante: {luis.estudiante}")
    print(f"Total a pagar: S/ {luis.costo_final}")

    rosa = Matricula("Rosa Ccahuana Prado")
    rosa.agregar_recargo(15.0, "Índice Desaprobatorio")
    rosa.agregar_recargo(15.0, "Extemporánea")
    rosa.aplicar_beca()

    print(f"Estudiante: {rosa.estudiante}")
    print(f"Total a pagar: S/ {rosa.costo_final}")


    print("\nvulnerar el sistema")
    try:
        # Alguien intenta modificar el costo base a mano para pagar menos
        luis.costo_base = 10.0
    except AttributeError as e:
        print(f"¡Operación bloqueada! El encapsulamiento funciona: {e}")
```
